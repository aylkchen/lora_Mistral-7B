```python
model_path = "/public/huggingface-models/mistralai/Mistral-7B-Instruct-v0.1"
```

```python
# 从 transformers 库中导入两个常用类：
# AutoTokenizer：负责把文本转换成模型能读懂的数字（token）
# AutoModelForCausalLM：负责加载“因果语言模型”，也就是能根据前文继续生成后文的模型
from transformers import AutoTokenizer, AutoModelForCausalLM

# 导入 PyTorch。这里虽然这段代码里暂时没直接用到 torch，
# 但后面做张量运算、推理、把数据放到 GPU 上时通常会用到
import torch


# =========================
# 1. 加载 tokenizer（分词器）
# =========================
tokenizer = AutoTokenizer.from_pretrained(
    model_path,      # model_path 是模型所在的路径，可以是本地文件夹，也可以是 Hugging Face 上的模型名
    use_fast=True    # 使用“快速版” tokenizer，一般速度更快，推荐开启
)


# =========================
# 2. 加载大语言模型
# =========================
model = AutoModelForCausalLM.from_pretrained(
    model_path,      # 从同一个路径加载模型参数
    device_map="auto" # 让程序自动决定把模型放到哪里运行：
                       # 如果有 GPU，通常会自动放到 GPU；
                       # 如果没有，就可能放到 CPU
)


# ============================================
# 3. 设置 pad_token_id（填充 token 的编号）
# ============================================
model.generation_config.pad_token_id = tokenizer.pad_token_id
```

```python
# 一条用户消息
messages = [
    {"role": "user", "content": "Who are you?"},
]

# 按聊天模型的格式处理消息，并转成模型可用的张量
inputs = tokenizer.apply_chat_template(
    messages,
    add_generation_prompt=True,   # 补一个“该助手回答了”的提示
    tokenize=True,                # 直接转成 token id
    return_dict=True,             # 返回字典
    return_tensors="pt",          # 返回 PyTorch 张量
).to(model.device)               # 放到模型所在设备上（GPU/CPU）

# 让模型生成回答，最多生成 40 个新 token
outputs = model.generate(**inputs, max_new_tokens=40)

# 只取模型新生成的那部分，不要前面的输入
response_ids = outputs[0][inputs["input_ids"].shape[-1]:]

# 把 token id 转回文本并打印
print(tokenizer.decode(response_ids))
```

```python
# 定义一条用户消息
messages = [
    {"role": "user", "content": "Tell me what is gravity?"},
]

# 按聊天模板处理消息，并转成模型输入张量
inputs = tokenizer.apply_chat_template(
    messages,
    add_generation_prompt=True,   # 提示模型：现在该 assistant 开始回答了
    tokenize=True,                # 把文本转成 token id
    return_dict=True,             # 返回字典格式，方便后面取值
    return_tensors="pt",          # 返回 PyTorch 张量
).to(model.device)               # 放到模型所在设备（GPU/CPU）

# 让模型生成回答，最多生成 40 个新 token
outputs = model.generate(**inputs, max_new_tokens=40)

# 取出模型新生成的部分（去掉前面的输入部分）
response_ids = outputs[0][inputs["input_ids"].shape[-1]:]

# 把生成的 token id 解码成文本并打印
print(tokenizer.decode(response_ids))
```

```python
import random  # 导入 random 库，这段代码里其实暂时没有用到

# 定义一个函数：把一条样本数据整理成聊天文本
def build_chat_text(example):
    # 按聊天格式构造消息列表
    messages = [
        {
            "role": "system",  # 系统提示词，告诉模型它要扮演什么角色
            "content": f"You are an anime character with the following personality: {example['trait']}."
            # 把样本里的 trait（性格描述）插入到提示词中
        },
        {
            "role": "user",    # 用户提问
            "content": example["question"]  # 取出样本里的问题
        },
        {
            "role": "assistant",  # 助手回答
            "content": example["dialogue"]  # 取出样本里的回答
        }
    ]

    # 用聊天模板把 messages 拼接成一段完整文本
    text = tokenizer.apply_chat_template(
        messages,
        tokenize=False,              # 不转成 token id，只返回普通字符串
        add_generation_prompt=False  # 不额外加“请开始回答”的提示
    )

    # 返回一个字典，里面只有一个字段 text
    return {"text": text}
```

```python
# 对 train_ds 里的每一条样本都执行 build_chat_text 函数
# 处理后生成一个新的数据集 dataset
dataset = train_ds.map(build_chat_text)

# 显示处理后的数据集信息
dataset
```

```python
# 从 peft 库中导入 LoRA 配置类和给模型加 LoRA 的函数
from peft import LoraConfig, get_peft_model

# 定义 LoRA 的配置
lora_config = LoraConfig(
    r=16,                       # LoRA 的低秩维度，越大可训练参数越多
    lora_alpha=32,              # LoRA 的缩放系数
    lora_dropout=0.05,          # LoRA 层的 dropout，防止过拟合
    target_modules=["q_proj", "k_proj", "v_proj", "o_proj"],
    # 指定把 LoRA 加到哪些层上
    # 这里是 attention 里的 q/k/v/o 投影层

    bias="none",                # 不训练原模型里的 bias 参数
    task_type="CAUSAL_LM"       # 任务类型：因果语言模型
)

# 把 LoRA 配置应用到原模型上
# 处理后，model 就变成了一个带 LoRA 适配器的模型
model = get_peft_model(model, lora_config)
```

```python
# 从 trl 库中导入 SFTTrainer
# SFTTrainer 是专门用来做监督微调（SFT）的训练器
from trl import SFTTrainer

# 从 transformers 库中导入 TrainingArguments
# TrainingArguments 用来设置训练时的各种参数
from transformers import TrainingArguments

# 定义训练参数
training_args = TrainingArguments(
    output_dir="./anime-mistral",      # 训练结果和模型检查点保存的文件夹

    per_device_train_batch_size=4,     # 每张 GPU 上一次喂给模型的样本数是 4

    gradient_accumulation_steps=4,     # 梯度累计 4 次再更新一次参数
    # 相当于把有效 batch size 变大
    # 这里等效于总共 4 × 4 = 16 个样本再更新一次

    learning_rate=2e-4,                # 学习率，控制每次参数更新的步子大小

    num_train_epochs=3,                # 训练轮数，总共把整个训练集学 3 遍

    fp16=True,                         # 使用半精度训练，可以减少显存占用、加快训练

    logging_steps=20,                  # 每训练 20 步打印一次日志

    save_steps=500,                    # 每训练 500 步保存一次模型

    save_total_limit=2,                # 最多只保留 2 个检查点，旧的会被删除

    report_to="none"                   # 不把训练日志上传到 wandb、tensorboard 等平台
)
```

```python
trainer = SFTTrainer(
    model=model,
    train_dataset=dataset,
    args=training_args
)
```

```python
trainer.train()
```

```python
import torch  # 导入 PyTorch

# 定义一个函数：让模型按照指定性格进行对话
def chat_with_personality(trait, user_input, max_new_tokens=100, temperature=0.8):
    """
    根据给定的性格 trait 和用户输入 user_input，生成回答
    """

    # 构造对话消息
    messages = [
        {
            "role": "system",  # 系统提示词：告诉模型要扮演什么性格
            "content": f"You are an anime character with the following personality: {trait}."
        },
        {
            "role": "user",    # 用户输入的问题
            "content": user_input
        }
    ]

    # 按聊天模板把消息拼成 prompt 字符串
    prompt = tokenizer.apply_chat_template(
        messages,
        tokenize=False,              # 这里只返回字符串，不直接转 token
        add_generation_prompt=True   # 在最后补一个“该 assistant 回答了”的提示
    )

    # 把 prompt 编码成模型能读的张量，并放到模型所在设备上
    inputs = tokenizer(prompt, return_tensors="pt").to(model.device)

    # 关闭梯度计算，表示这里只做推理，不做训练
    with torch.no_grad():
        outputs = model.generate(
            **inputs,                         # 把输入传给模型
            max_new_tokens=max_new_tokens,   # 最多生成多少个新 token
            do_sample=True,                  # 使用采样，而不是每次都选概率最大的 token
            temperature=temperature,         # 控制生成的随机性，越大越随机
            top_p=0.9,                       # nucleus sampling，只从累计概率前 90% 的候选里采样
            repetition_penalty=1.1,          # 重复惩罚，减少重复输出
            pad_token_id=tokenizer.eos_token_id  # 设置 padding token id，这里用 eos token
        )

    # 把输出 token 解码成文本
    response = tokenizer.decode(outputs[0], skip_special_tokens=True)

    # 去掉前面的 prompt，只保留模型真正生成的回答
    if "[/INST]" in response:
        response = response.split("[/INST]")[-1].strip()

    # 返回最终回答
    return response
```

```python
print("---- TSUNDERE (傲娇) ----")
tsundere = chat_with_personality("tsundere", query)
print(tsundere)

print("\n---- YANDERE (病娇) ----")
yandere = chat_with_personality("yandere", query)
print(yandere)

print("\n---- BAKADERE (笨蛋娇) ----")
bakadere = chat_with_personality("bakadere", query)
print(bakadere)

print("\n---- HIMEDERE (公主娇) ----")
himedere = chat_with_personality("himedere", query)
print(himedere)

print("\n---- GENKI (元气型) ----")
genki = chat_with_personality("genki", query)
print(genki)

print("\n---- MOE (萌系 ) ----")
moe = chat_with_personality("moe", query)
print(moe)

```
